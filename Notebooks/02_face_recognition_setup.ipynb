# Face Recognition Setup & Training

This notebook trains the face recognition embeddings database from customer sample face images and saves the `face_db.pkl` model file.

### Step 1: Load Data & Initialize Module

```python
import os
import cv2
import pickle
import numpy as np
import sys
sys.path.append('..')
from face_recognition_module import FaceRecognitionModule

# Create sample customer dataset directory if not present
base_cust_dir = 'customers'
os.makedirs(base_cust_dir, exist_ok=True)

customers = ['CUST001', 'CUST002', 'CUST003']
for cust in customers:
    cust_path = os.path.join(base_cust_dir, cust)
    os.makedirs(cust_path, exist_ok=True)
    for img_idx in range(1, 4):
        file_name = os.path.join(cust_path, f"img_{img_idx}.jpg")
        if not os.path.exists(file_name):
            img = np.full((200, 200, 3), 220, dtype=np.uint8)
            cv2.ellipse(img, (100, 100), (60, 80), 0, 0, 360, (180, 180, 180), -1)
            eye_color = (50 * int(cust[-1]), 100, 150)
            cv2.circle(img, (75, 80), 12, eye_color, -1)
            cv2.circle(img, (125, 80), 12, eye_color, -1)
            cv2.ellipse(img, (100, 130), (30, 15), 0, 0, 180, (50, 50, 50), 3)
            cv2.imwrite(file_name, img)

print(f"Customer dataset ready in '{base_cust_dir}/'.")
```

### Step 2: Train Model & Build Face Embeddings Database

```python
# Initialize module
fr_module = FaceRecognitionModule(db_path='face_db.pkl')

# Build face database encodings from dataset
face_db = fr_module.build_database(customer_dir=base_cust_dir)
print(f"Trained embeddings for {len(face_db)} registered customers.")
```

### Step 3: Save Model Artifact (`face_db.pkl`)

```python
# Save trained encodings database to face_db.pkl
fr_module.save_database()
print("Saved 'face_db.pkl' model file successfully!")
```
